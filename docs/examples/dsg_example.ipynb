# Dynamic Pipeline — `DynamicScenarioGenerator`

Extends the FM and AF baselines **beyond 2070** using **NAI-based (Net Annual Increment) harvest**, and extends the scenario end year to match. Mirrors `tests/examples/dsg_example.py`.

```python
import pandas as pd
from goblin_cbm_runner.dynamic_scenario_generator import DynamicScenarioGenerator

scenario_df = pd.read_csv('../../tests/examples/data/input/scenario_dataframe.csv')
afforest_df = pd.read_csv('../../tests/examples/data/input/cbm_afforestation.csv', index_col=0)
```

## Run the extended simulation

`dynamic_config` overrides the `DF_config.yaml` defaults. `harvest_ratio=0.75` targets a sustainable 75% of annual growth; `dynamic_years` sets how far past 2070 to continue.

```python
dsg = DynamicScenarioGenerator(
    scenario_data=scenario_df,
    afforestation_data=afforest_df,
    config={'afforest_delay': 5, 'annual_rate_pre_delay': 1200},
    dynamic_config={'harvest_ratio': 0.75, 'dynamic_years': 50},
)

results = dsg.run_flux_simulation()        # extended FM + AF + SC
results.head()
```

## Baseline only

`run_baseline_flux_simulation()` returns the extended FM + AF baseline without the SC scenarios.

```python
baseline = dsg.run_baseline_flux_simulation()
baseline.head()
```
