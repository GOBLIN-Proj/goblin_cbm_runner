# Standard Pipeline (user CSVs) — `StandardSimGenerator`

Run a simulation from your **own inventory CSV files** with an explicit disturbance schedule. No internal database of stands is required (the bundled AIDB is still used for CBM parameters). Mirrors `tests/examples/standard_sim_example.py`.

## Get the template

Copy the bundled template CSVs, then edit them for your forest (inventory, classifiers, age_classes, disturbance_types, disturbance_events, transitions, growth, sit_config.json).

```python
from goblin_cbm_runner.standard_sim_generator import StandardSimGenerator

StandardSimGenerator.get_template('./my_forest/')
```

## Run the simulation

`baseline_year` and `end_year` set the static simulation window; the disturbance schedule comes from your `disturbance_events.csv`.

```python
gen = StandardSimGenerator(
    csv_directory='./my_forest/',
    config={'baseline_year': 2020, 'end_year': 2050},
    scenario=0,
)

results = gen.run_flux_simulation()
results.head()
```
